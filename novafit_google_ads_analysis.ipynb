```python
numeric_columns = [
    'Impressions',
    'Clicks',
    'Conversions',
    'Quality_Score'
]

df_clean[numeric_columns] = df_clean[numeric_columns].apply(
    pd.to_numeric,
    errors='coerce'
)
```

```python
df_clean['CTR'] = np.where(
    df_clean['Impressions'] > 0,
    (df_clean['Clicks'] / df_clean['Impressions']) * 100,
    0
)

df_clean['CPC'] = np.where(
    df_clean['Clicks'] > 0,
    df_clean['Cost_COP'] / df_clean['Clicks'],
    0
)

df_clean['CPA'] = np.where(
    df_clean['Conversions'] > 0,
    df_clean['Cost_COP'] / df_clean['Conversions'],
    0
)

df_clean['Conversion_Rate'] = np.where(
    df_clean['Clicks'] > 0,
    (df_clean['Conversions'] / df_clean['Clicks']) * 100,
    0
)
```

```python
campaign_type_summary = df_clean.groupby('Campaign_Type').agg({
    'Impressions': 'sum',
    'Clicks': 'sum',
    'Cost_COP': 'sum',
    'Conversions': 'sum'
}).reset_index()

campaign_type_summary
```

```python
campaign_type_summary['CTR'] = (
    campaign_type_summary['Clicks'] /
    campaign_type_summary['Impressions']
) * 100

campaign_type_summary['CPC'] = (
    campaign_type_summary['Cost_COP'] /
    campaign_type_summary['Clicks']
)

campaign_type_summary['CPA'] = (
    campaign_type_summary['Cost_COP'] /
    campaign_type_summary['Conversions']
)

campaign_type_summary['Conversion_Rate'] = (
    campaign_type_summary['Conversions'] /
    campaign_type_summary['Clicks']
) * 100

campaign_type_summary
```

**Campaign Performance Analysis**

The analysis shows clear differences in performance across NovaFit's three campaign types: Brand, Category and Competitor.

### Brand Campaign

The Brand campaign was the strongest-performing campaign overall. It generated 396 conversions from 36,530 impressions and 3,858 clicks, while spending 3,700,915 COP. It achieved the highest CTR (10.56%), lowest CPC (959.28 COP) and lowest CPA (9,345.74 COP) of the three campaigns. Its 10.26% conversion rate was also substantially higher than Category and Competitor.

These results suggest that users searching for NovaFit-related terms were highly engaged and more likely to convert, making Brand the most efficient campaign in the dataset.

### Category Campaign

The Category campaign generated the highest traffic volume, receiving 144,730.59 impressions and 7,646 clicks. It also accounted for the largest amount of spend at 14,374,023 COP and generated 257 conversions.

However, its efficiency was considerably lower than Brand. Category achieved a 5.28% CTR, 1,879.94 COP CPC, 55,930.05 COP CPA and 3.36% conversion rate. Although it generated a significant number of conversions, the much higher CPA indicates that these conversions were acquired at a substantially greater cost than Brand conversions.

This suggests that Category provides valuable acquisition potential but may require optimisation to improve conversion efficiency.

### Competitor Campaign

The Competitor campaign performed the weakest overall. It generated 27,475 impressions and 314 clicks while spending 992,595 COP, but produced zero conversions. Its 1.14% CTR was the lowest of all three campaigns, while its 3,161.13 COP CPC was the highest.

As there were no conversions, a CPA cannot be calculated meaningfully. The 0% conversion rate further indicates that the traffic generated by the Competitor campaign did not result in conversions during the period analysed.

### Overall Findings

Overall, Brand was the most efficient campaign, combining the highest conversion volume with the lowest CPC and CPA and the highest CTR and conversion rate. Category generated the greatest amount of traffic and spend, but its substantially higher CPA indicates that there is room to improve efficiency. Competitor was the weakest campaign, generating clicks but no conversions.

From a strategic perspective, the results suggest that NovaFit should protect the strong performance of Brand, optimise Category to improve its conversion efficiency, and carefully review the continued allocation of spend to Competitor given its lack of conversions.

## Key Findings

- **Brand was the most efficient campaign**, achieving the highest CTR (10.56%), highest conversion rate (10.26%), and lowest CPA (9,345.74 COP).
- **Category generated the greatest volume**, with 144,730.59 impressions, 7,646 clicks and 257 conversions, but had a substantially higher CPA of 55,930.05 COP.
- **Competitor generated no conversions** despite spending 992,595 COP, with the lowest CTR (1.14%) and highest CPC (3,161.13 COP).
- The results indicate that Brand currently provides the strongest return from search activity, while Category represents an opportunity for optimisation and Competitor requires careful budget consideration.

## Recommendations

### 1. Protect Brand performance

Brand should continue to receive sufficient budget to capture high-intent searches because it generated the highest number of conversions at the lowest CPA.

### 2. Optimise Category performance

Category generates substantial traffic and conversions but at a much higher CPA than Brand. Keyword, ad and landing-page performance should therefore be reviewed to identify opportunities to improve conversion efficiency.

### 3. Review Competitor spend

The Competitor campaign generated 314 clicks but no conversions. Given its 0% conversion rate and highest CPC, its budget should be carefully reviewed before increasing spend.

### 4. Monitor campaign efficiency

Future optimisation should consider CPA, conversion rate, CTR and CPC together rather than relying on conversion volume alone. This would help NovaFit balance customer acquisition volume with cost efficiency.
